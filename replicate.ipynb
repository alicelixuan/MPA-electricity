# Replication: all figures and tables

**Run all cells.** With the settings below (all `False`), every figure and table of the paper is produced from the stored data in about a minute, written to `output/paper/` and shown below.

To recompute a step instead of reading its stored result, set its flag to `True`. The step and everything downstream are written to `output/rerun/<run>/` with `comparison.md`; the paper outputs below are always the stored-data ones. Reruns are random (seeded) and do not reproduce the stored results exactly; see `README.md`.

| Step | Stored result | Flag | Time if rerun |
|---|---|---|---|
| 1. Supply regimes (PELT) | `data/stored/change_points.json` | `RERUN_REGIMES` | ~1 s |
| 2. Supply-curve fits, hourly slopes | `data/stored/supply_fits.csv`, `slopes.parquet` | `RERUN_FITS` | 20–35 min (2 cores) |
| 3. Monte Carlo dispatch model | `data/stored/dispatch_results.parquet` | `RERUN_DISPATCH_MODEL` | ~230 CPU-hours |
| 4. Analysis panel | `data/stored/analysis_panel.parquet` | `RERUN_PANEL` | seconds |
| 5–6. Econometrics, figures, tables | – | always | ~1 min |

## Settings

```python
RERUN_REGIMES = False          # step 1; new regimes force a refit
RERUN_FITS = False             # step 2
RERUN_DISPATCH_MODEL = False   # step 3; needs pyomo and a solver (requirements-model.txt)
RERUN_PANEL = False            # step 4; forced on by any rerun above

SEED = 2024                    # fits and dispatch model
FIT_RESTARTS = 1000            # random restarts per fit, as stored
MC_ITERATIONS = 1000           # Monte Carlo iterations, as stored
SOLVER = "glpk"                # "glpk" (original runs) or "appsi_highs" (pip install highspy)
N_JOBS = 1                     # parallel processes
RUN_NAME = None                # folder under output/rerun/; None = date and time
DISPATCH_UNITS = None          # optional list of EICs: partial dispatch rerun (others keep stored results)
```

## Run

```python
from mpa import pipeline
from mpa.report import show, show_rerun

result = pipeline.run(RERUN_REGIMES, RERUN_FITS, RERUN_DISPATCH_MODEL, RERUN_PANEL, SEED, FIT_RESTARTS,
                      MC_ITERATIONS, SOLVER, N_JOBS, RUN_NAME, DISPATCH_UNITS)
```
